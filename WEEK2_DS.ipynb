**PART 1:BUSINESS UNDERSTANDING**


## Heart Disease Prediction

### 1. Business Problem

Cardiovascular diseases are a major global health concern and are the leading cause of death worldwide. According to the World Health Organization (WHO), an estimated 19.8 million people died from cardiovascular diseases in 2022. Early detection and appropriate management can help reduce their impact on individuals and healthcare systems.

Healthcare professionals consider several factors when assessing a patient's heart health, including age, sex, blood pressure, cholesterol levels, chest pain characteristics, and results from cardiac examinations. Analysing these factors together can be challenging, particularly when large amounts of patient information must be evaluated.

The business problem addressed in this project is how to use patient health information to support the identification of patients who may have heart disease. A machine learning approach can analyse patterns in historical patient records and learn relationships between clinical measurements and recorded heart disease outcomes.

### 2. Project Objective

The objective of this project is to prepare the Heart Failure Prediction dataset for a machine learning classification task that predicts whether heart disease is present based on the available patient information.

The project involves understanding the dataset, examining its quality, identifying missing or potentially invalid values, detecting outliers, performing exploratory data analysis, encoding categorical variables, evaluating feature relationships, and preparing a final dataset suitable for machine learning.

The analysis will also investigate which features may be relevant to prediction, which variables require encoding or scaling, and whether highly correlated or redundant features need further consideration.

The overall aim is to produce a clean, consistent, and reliable dataset that can be used to develop and evaluate a classification model.

### 3. Target Variable

The target variable is **HeartDisease**, which indicates the recorded heart disease outcome for each patient.

- **1:** Heart disease is present.
- **0:** Heart disease is absent.

The dataset contains 918 patient records and 11 predictor features, including demographic information and clinical measurements. These features include age, sex, chest pain type, resting blood pressure, cholesterol, fasting blood sugar, resting electrocardiogram results, maximum heart rate, exercise-induced angina, Oldpeak, and the slope of the peak exercise ST segment.

Since the target variable has two possible values, this is a **supervised binary classification problem**. The model will learn from labelled examples to predict the target class for new records.

Importantly, the dataset represents recorded outcomes rather than following patients over time. Therefore, this project focuses on predicting the presence of heart disease as represented in the dataset, rather than predicting future disease onset or providing a clinical diagnosis.

### 4. Importance of Predictive Modelling

Predictive modelling can help identify patterns in patient data that may be difficult to recognise by examining individual measurements separately. Machine learning algorithms can evaluate multiple features simultaneously and estimate which patterns are associated with the recorded heart disease outcome.

This approach is valuable because heart disease assessment may involve several interacting factors. Analysing these factors together may help support more systematic screening and further clinical assessment.

However, predictive models must be evaluated carefully using appropriate performance measures, including accuracy, precision, recall, F1-score, and a confusion matrix. In a healthcare context, recall is particularly relevant because false-negative predictions may fail to flag patients whose records indicate heart disease. Model performance, limitations, and the consequences of different errors must therefore be considered before any real-world use.

A model developed using this dataset should be treated as an analytical and educational tool, not as a replacement for professional medical judgement.

### 5. Expected Business Impact

The expected benefits of this project include:

- **Improved data quality:** Identifying and addressing missing, invalid, inconsistent, and potentially problematic values before model development.
- **Better understanding of relevant features:** Exploring which demographic and clinical variables are associated with the recorded heart disease outcome.
- **Support for screening research:** Demonstrating how machine learning could assist in prioritising records for further review, subject to appropriate clinical validation.
- **More consistent analysis:** Applying a systematic data preparation process that can be repeated and evaluated.
- **A foundation for future development:** Producing a machine learning-ready dataset that can be used to train, test, and compare classification algorithms.

In conclusion, this project applies data analysis and preprocessing techniques to prepare a heart disease dataset for predictive modelling. The expected outcome is a high-quality dataset that supports further investigation into heart disease classification. Any practical healthcare benefit would depend on additional validation, representative data, and appropriate clinical oversight.

OUTLIER DETECTION

```python
# Boxplot
import matplotlib.pyplot as plt
import seaborn as sns

num_cols = ['age', 'resting_bp', 'cholesterol', 'max_hr', 'old_peak']

plt.figure(figsize=(12, 6))

for i, col in enumerate(num_cols, 1):
    plt.subplot(2, 3, i)
    sns.boxplot(y=df_fe[col])
    plt.title(col)

plt.tight_layout()
plt.show()
```

```python
# IQR method
for col in num_cols:
    Q1 = df_fe[col].quantile(0.25)
    Q3 = df_fe[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    count = ((df_fe[col] < lower) |
             (df_fe[col] > upper)).sum()

    print(f"{col}: {count} potential outliers")
```

```python
# z-score method
from scipy.stats import zscore

for col in num_cols:
    z_scores = zscore(df_fe[col])
    count = (abs(z_scores) > 3).sum()

    print(f"{col}: {count} potential outliers")
```

### Outlier Detection and Handling

Outlier detection was performed on the continuous numerical features: age, resting blood pressure, cholesterol, maximum heart rate, and oldpeak.

Boxplots were used to visualize potential outliers. The Interquartile Range (IQR) method was applied using the 1.5 × IQR rule, and the Z-score method was used to identify observations more than three standard deviations from the mean.

Since the dataset contains medical measurements, potential outliers were not automatically removed. Unusual values may represent genuine patient conditions and could contain useful predictive information. Therefore, the detected observations should be investigated before deciding whether correction, removal, or retention is appropriate.

```python
# inspecting the actual values
for col in num_cols:
    Q1 = df_fe[col].quantile(0.25)
    Q3 = df_fe[col].quantile(0.75)
    IQR = Q3 - Q1

    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR

    outliers = df_fe.loc[
        (df_fe[col] < lower) | (df_fe[col] > upper), col
    ]

    print(f"\n{col}")
    print("Lower bound:", lower)
    print("Upper bound:", upper)
    print("Potential outlier values:")
    print(sorted(outliers.unique()))
```

```python
# Check suspicious zero values
print("Resting BP = 0:", (df_fe['resting_bp'] == 0).sum())
print("Cholesterol = 0:", (df_fe['cholesterol'] == 0).sum())
```

```python
# Handling invalid values
import numpy as np
from sklearn.impute import SimpleImputer

# Mark suspicious zero values as missing
df_fe.loc[df_fe['resting_bp'] == 0, 'resting_bp'] = np.nan
df_fe.loc[df_fe['cholesterol'] == 0, 'cholesterol'] = np.nan

# Fill missing values using the median
imputer = SimpleImputer(strategy='median')

cols_to_impute = ['resting_bp', 'cholesterol']

df_fe[cols_to_impute] = imputer.fit_transform(
    df_fe[cols_to_impute]
)

# Verify
print(df_fe[cols_to_impute].isnull().sum())
print("Dataset shape:", df_fe.shape)
```

```python
# The remaining potential outliers were retained because
# extreme medical measurements may be genuine and relevant to heart disease prediction.
# No additional rows were removed solely on the basis of statistical thresholds.
```



The final dataset was prepared for subsequent machine learning tasks by applying the following preprocessing steps:

1. **Column standardization:** Column names were converted to lowercase snake_case format to improve consistency and ease of use in Python.

2. **Missing and invalid values:** Zero values in `resting_bp` and `cholesterol` were treated as potentially invalid measurements and replaced with missing values. These missing values were then imputed using the median of the respective columns.

3. **Outlier analysis:** The IQR method and Z-score method were used to identify potential outliers in numerical features. Remaining outliers were retained because extreme medical measurements may represent genuine observations and should not be removed without justification.

4. **Categorical encoding:** One-hot encoding was applied to `chest_pain_type`, `resting_ecg`, and `st_slope`. Binary encoding was applied to `sex` and `exercise_angina`.

5. **Feature selection:** Correlation analysis and a correlation heatmap were used to examine relationships between predictor variables and the target variable, `heart_disease`. Features were not removed solely because of weak correlation.

6. **Feature engineering:** The engineered `age_group` column was excluded from the initial final dataset because the original numerical `age` feature was retained.

7. **Target separation:** The `heart_disease` column was retained as the prediction target, with 0 and 1 representing the two target classes.

8. **Final validation:** The final dataset was checked for missing values, data types, duplicate rows, and target values before machine learning.

The resulting dataset contains numerical predictor features and a binary target variable, making it suitable for use in classification models. Feature scaling and model-specific preprocessing can be applied within a training pipeline when required.
